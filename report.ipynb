# Part I: Model Development

### 1. Identification and Definition of Key Input Features

The performance of a human fighter pilot is influenced by multiple interacting physical, cognitive, and environmental factors. In this model, five key input features are identified as the primary determinants of fighter pilot performance. All input variables are normalized within the range [0,1] to ensure consistency and comparability.

- Heart Rate (HR):

Heart rate represents the pilot’s physiological stress level during mission execution. Elevated heart rate values indicate increased physical exertion or psychological stress, which may negatively affect reaction time, decision-making accuracy, and cognitive efficiency.

- Sleep Quality (SQ):
  
Sleep quality reflects the level of rest and recovery achieved by the pilot prior to the mission. Low sleep quality is associated with fatigue, reduced alertness, slower reaction times, and increased cognitive load. Higher sleep quality improves mental resilience and sustained performance.

- Mission Complexity (MC):
  
Mission complexity represents the level of cognitive demand required to successfully complete the mission. Complex missions involve multitasking, rapid information processing, and dynamic decision-making, all of which increase mental workload and stress.

- Experience Level (EL):
  
Experience level indicates the number of years or intensity of flight training and operational exposure. More experienced pilots are generally better at managing stress, maintaining situational awareness, and performing consistently under high workload conditions.

- Environmental Stress (ES):
  
Environmental stress accounts for adverse operational conditions such as poor weather, low visibility, turbulence, and strong winds. These factors increase both physical and cognitive demands on the pilot and can degrade navigation accuracy and aircraft control.

### 2. Mathematical Representation of Input Features

To maintain a unified modeling framework, all input features are normalized to values between 0 and 1. This normalization allows the model to remain dimensionless and simplifies the interaction between variables.

Let the normalized inputs be defined as:

- HR,SQ,MC,EL,ES ∈ [0,1]


Where:

0 represents the minimum possible influence,

1 represents the maximum possible influence.

Normalization enables the model to remain scalable and adaptable to different datasets and operational conditions without changing the underlying structure.

### 3. Dynamics and Interplay of Internal Variables

The internal dynamics of the model describe how the input features interact to influence pilot performance. These interactions occur at both instantaneous and conceptual levels.

##### 3.1 Instantaneous Interactions

Instantaneous interactions represent relationships where changes in input variables immediately affect internal system states. For example:

An increase in mission complexity or environmental stress directly raises cognitive load.

A high heart rate instantly contributes to physical fatigue.

Reduced sleep quality immediately lowers cognitive resilience.

These effects are modeled without time delay, assuming that the pilot’s physiological and cognitive responses occur rapidly during mission execution.

##### 3.2 Conceptual (Aggregated) Interactions

Aggregated interactions describe how multiple internal variables combine to influence higher-level performance outcomes:

Cognitive load, physical fatigue, and environmental stress collectively increase overall stress level.

Experience level and sleep quality mitigate the negative impact of cognitive load by enhancing situational awareness.

Pilot performance emerges as a balance between positive contributors (situational awareness) and negative contributors (stress level).

This layered interaction structure reflects real-world pilot behavior, where multiple stressors and coping mechanisms operate simultaneously.

### 4. Mathematical Specifications of Internal Variables
##### 4.1 Identification of Internal Variables

The model defines the following internal variables:

Cognitive Load (CL): Mental effort required to process mission-related information.

Physical Fatigue (PF): Level of physical exhaustion resulting from stress and exertion.

Stress Level (SL): Combined psychological and physiological stress experienced by the pilot.

Situational Awareness (SA): Pilot’s ability to perceive, comprehend, and anticipate mission conditions.

Pilot Performance (PP): Overall effectiveness of the pilot during the mission.

All internal variables are also normalized within the range [0,1].

##### 4.2 Mathematical Equations

The internal variables are mathematically specified as follows:

- Cognitive Load
CL=0.3⋅MC+0.3⋅ES+0.3⋅HR−0.3⋅EL

This equation reflects the increase in cognitive load due to mission complexity, environmental stress, and physiological stress, while experience level reduces the perceived mental effort.

- Physical Fatigue
PF=0.5⋅HR−0.4⋅SQ

Higher heart rate increases fatigue, while good sleep quality reduces physical exhaustion.

- Stress Level
SL=0.3⋅CL+0.3⋅PF+0.33ES-0.1EL

Stress level is influenced by both mental and physical fatigue, with environmental stress acting as an additional stressor. while experience level reduces the stress levels due to added confidence.

- Situational Awareness
SA=0.7⋅EL+0.7⋅SQ−0.4⋅CL

Situational awareness improves with experience and adequate sleep, while high cognitive load degrades the pilot’s ability to process and interpret information effectively.

- Pilot Performance
PP=0.8⋅SA−0.4⋅SL

Pilot performance is modeled as a trade-off between situational awareness and stress. Higher awareness enhances performance, whereas excessive stress degrades mission effectiveness.

##### 4.3 Parameter Justification

The coefficients used in the equations represent relative influence weights based on domain knowledge:

Larger coefficients indicate stronger influence.

Negative coefficients represent mitigating or compensatory effects.

The model structure emphasizes realism rather than exact physiological measurement, making it suitable for simulation, sensitivity analysis, and training applications.

All computed values are constrained to the range [0,1] to preserve model stability and interpretability.

# Part II: Simulation
### 1. Simulation Scenarios

To evaluate the behavior of the fighter pilot performance model under different operational conditions, a set of six simulation scenarios was designed. Each scenario represents a realistic mission profile with varying levels of physical, cognitive, and environmental demands. All input values are normalized within the range [0,1].

- Scenario 1: Ideal Mission Conditions

This scenario represents an optimal operational state where the pilot is well-rested, experienced, and operating in favorable environmental conditions with low mission complexity. Minimal stress and fatigue are expected, resulting in high performance.

- Scenario 2: High-Stress Combat Mission

This scenario models an intense combat situation characterized by high mission complexity, elevated heart rate, adverse environmental conditions, and moderate experience. The pilot is expected to experience high cognitive load and stress, which will negatively affect performance.

- Scenario 3: Fatigued Pilot

This scenario represents a pilot suffering from poor sleep quality prior to the mission. Although mission complexity and environmental conditions are moderate, accumulated fatigue is expected to significantly reduce situational awareness and overall performance.

- Scenario 4: Inexperienced Pilot

This scenario examines the effect of low experience level on performance. Despite average sleep quality and manageable environmental conditions, the lack of experience increases cognitive load and reduces the pilot’s ability to handle mission demands efficiently.

- Scenario 5: Adverse Environmental Conditions

This scenario focuses on the impact of severe environmental stress, such as poor weather or low visibility. Even with reasonable sleep quality and experience, environmental stress increases cognitive and physical demands, reducing performance.

- Scenario 6: Experienced Pilot Under Extreme Pressure

This scenario represents a highly experienced pilot facing a highly complex mission under stressful environmental conditions. While experience mitigates some negative effects, extreme mission demands still impose high stress levels.

### 2. Experimental Scenario Setup

The simulation is conducted by assigning a predefined set of normalized input values to each scenario. The model processes these inputs using the mathematical equations defined in Part I to compute the internal variables and final pilot performance.

The experimental setup follows these steps:

Define normalized input vectors for each scenario.

Compute internal variables, including cognitive load, physical fatigue, stress level, and situational awareness.

Calculate the resulting pilot performance.

Compare outputs across scenarios to assess model behavior.

This structured approach ensures repeatability and consistency across all simulation runs.

### 4. Python Implementation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
## utility function (ensure the values are within the correct range)
def clamp(x):
    return np.clip(x, 0, 1)
```

```python
def fighter_pilot_model(HR, SQ, MC, EL, ES):
    # internal variables
    cognitive_load = clamp(
        0.3 * MC + 0.3 * ES + 0.3 * HR - 0.3 * EL
    )

    physical_fatigue = clamp(
        0.5 * HR - 0.4 * SQ
    )

    stress_level = clamp(
        0.3 * cognitive_load + 0.3 * physical_fatigue + 0.3 * ES -0.1 * EL
    )

    situational_awareness = clamp(
        0.7 * EL + 0.7 * SQ - 0.4 * cognitive_load
    )

    pilot_performance = clamp(
        0.8 * situational_awareness - 0.4 * stress_level
    )
    return {
        "Cognitive Load": cognitive_load,
        "Physical Fatigue": physical_fatigue,
        "Stress Level": stress_level,
        "Situational Awareness": situational_awareness,
        "Pilot Performance": pilot_performance
    }
```

```python
scenarios = {
    "Ideal Mission":              [0.3, 0.9, 0.3, 0.8, 0.2],
    "High-Stress Combat":         [0.8, 0.4, 0.9, 0.4, 0.9],
    "Fatigued Pilot":             [0.7, 0.2, 0.6, 0.5, 0.3],
    "Inexperienced Pilot":        [0.6, 0.6, 0.7, 0.1, 0.4],
    "Adverse Environment":        [0.5, 0.6, 0.6, 0.6, 0.9],
    "Experienced Under Pressure": [0.8, 0.6, 0.9, 0.9, 0.8]
}
```

```python
results = {}

for scenario, values in scenarios.items():
    HR, SQ, MC, EL, ES = values
    results[scenario] = fighter_pilot_model(HR, SQ, MC, EL, ES)

results_df = pd.DataFrame(results).T
results_df
```

```python
results_df.plot(
    kind="bar",
    figsize=(14, 6)
)

plt.title("Simulation Results of Fighter Pilot Performance Model")
plt.ylabel("Normalized Value")
plt.xlabel("Simulation Scenarios")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
inputs_df = pd.DataFrame(
    scenarios,
    index=["Heart Rate", "Sleep Quality", "Mission Complexity",
           "Experience Level", "Environmental Stress"]
).T

inputs_df.plot(
    kind="bar",
    figsize=(14, 6)
)

plt.title("Input Feature Distribution Across Scenarios")
plt.ylabel("Normalized Value")
plt.xlabel("Scenario")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
results_df["Pilot Performance"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Pilot Performance Across Simulation Scenarios")
plt.ylabel("Performance Level")
plt.xlabel("Scenario")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# Part III: Evaluation
### 1. Model Verification

Model verification aims to ensure that the mathematical formulation and implementation of the fighter pilot performance model are logically correct and internally consistent.

##### 1.1 Verification Techniques

Several verification techniques will be applied to assess the correctness of the model:

Boundary testing: All input variables will be tested at their minimum and maximum values to ensure that the model remains numerically stable and that all outputs remain within the normalized range [0,1].

Logical consistency testing: Input variables will be varied individually while keeping others constant to verify that the model responds in a manner consistent with real-world expectations.

Extreme-case analysis: Scenarios representing ideal and worst-case conditions will be evaluated to confirm that the model produces realistic performance outcomes.

These techniques will confirm that the model implementation accurately reflects the intended mathematical structure.
##### Implementation

```python
# Boundary Testing

# All inputs at minimum
low_case = fighter_pilot_model(0, 0, 0, 0, 0)

# All inputs at maximum
high_case = fighter_pilot_model(1, 1, 1, 1, 1)

print("All inputs = 0:", low_case)
print("All inputs = 1:", high_case)
```

```python

test_values = np.linspace(0, 1, 20)

variables = [
    ("Heart Rate", "HR"),
    ("Sleep Quality", "SQ"),
    ("Mission Complexity", "MC"),
    ("Experience Level", "EL"),
    ("Environmental Conditions", "ES") 
]


fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

# Baseline values
base = {"HR": 0.5, "SQ": 0.5, "MC": 0.5, "EL": 0.5, "ES": 0.5}

# Iterate through each variable to generate plots
for i, (label, key) in enumerate(variables):
    performance_results = []
    
    for val in test_values:
        # Create a temporary parameter set with one variable changed
        current_params = base.copy()
        current_params[key] = val
        
        out = fighter_pilot_model(**current_params)
        performance_results.append(out["Pilot Performance"])
    
    # Plotting on the specific subplot index
    axes[i].plot(test_values, performance_results, color='steelblue', linewidth=2)
    axes[i].set_title(f"Impact of {label}")
    axes[i].set_xlabel(f"{label} (0 to 1)")
    axes[i].set_ylabel("Performance")
    axes[i].grid(True, linestyle='--', alpha=0.6)
    

fig.delaxes(axes[5])

plt.suptitle("Model Verification: Individual Variable Impact on Performance", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
